```python
import pandas as pd
import numpy as np
```

```python
path = "/home/sirsimsius/Dokumente/Arbeit/KU/GEMLST/GEMLST_MODIS/Data/SKL_Coefficients_Land_comb.csv"
land = pd.read_csv(path)
ice = pd.read_csv("/home/sirsimsius/Dokumente/Arbeit/KU/GEMLST/GEMLST_MODIS/Data/SKL_Coefficients_Ice_comb.csv")
```

```python
# Filter the land df for all rows that have or have not 'MOD_Terra_DAY' or 'MOD_Terra_NIGHT' in the 'Combination' column (in any combination)
all_mod_terra = land[land['Combination'].str.contains('MOD_Terra_DAY|MOD_Terra_NIGHT')]
no_mod_terra = land[~land['Combination'].str.contains('MOD_Terra_DAY|MOD_Terra_NIGHT')]
all_land = land['n_obs'].sum()

all_mod_terra_ice = ice[ice['Combination'].str.contains('MOD_Terra_DAY|MOD_Terra_NIGHT')]
no_mod_terra_ice = ice[~ice['Combination'].str.contains('MOD_Terra_DAY|MOD_Terra_NIGHT')]
all_ice = ice['n_obs'].sum()

# Print sum
print("LAND")
print("Sum of all MODIS Terra coefficients:", all_mod_terra['n_obs'].sum())
print("Sum of all non-MODIS Terra coefficients:", no_mod_terra['n_obs'].sum())
print("Percentual increase:", (all_land- all_mod_terra['n_obs'].sum()) / all_mod_terra['n_obs'].sum() * 100)

print("\nICE")
print("Sum of all MODIS Terra coefficients:", all_mod_terra_ice['n_obs'].sum())
print("Sum of all non-MODIS Terra coefficients:", no_mod_terra_ice['n_obs'].sum())
print("Percentual increase:", (all_ice - all_mod_terra_ice['n_obs'].sum()) / all_mod_terra_ice['n_obs'].sum() * 100)

```
