```python
import pandas as od
import numpy as np
import rasterio as rio
```

```python
# Split LST monthly and find lst trends with p<0.05. 

path = "/home/sirsimsius/Dokumente/Arbeit/KU/GIS/Output_Trends/lst_trend_monthly_seasonal.tif"
with rio.open(path) as src:
    jan_img = src.read(1)
    feb_img = src.read(2)
    mar_img = src.read(3)
    apr_img = src.read(4)
    may_img = src.read(5)
    jun_img = src.read(6)
    jul_img = src.read(7)
    aug_img = src.read(8)
    sep_img = src.read(9)
    oct_img = src.read(10)
    nov_img = src.read(11)
    dec_img = src.read(12)
    jan_pval = src.read(13)
    feb_pval = src.read(14)
    mar_pval = src.read(15)
    apr_pval = src.read(16)
    may_pval = src.read(17)
    jun_pval = src.read(18)
    jul_pval = src.read(19)
    aug_pval = src.read(20)
    sep_pval = src.read(21)
    oct_pval = src.read(22)
    nov_pval = src.read(23)
    dec_pval = src.read(24)

jan_msk = np.where(jan_pval < 0.05, jan_img, np.nan)
feb_msk = np.where(feb_pval < 0.05, feb_img, np.nan)
mar_msk = np.where(mar_pval < 0.05, mar_img, np.nan)
apr_msk = np.where(apr_pval < 0.05, apr_img, np.nan)
may_msk = np.where(may_pval < 0.05, may_img, np.nan)
jun_msk = np.where(jun_pval < 0.05, jun_img, np.nan)
jul_msk = np.where(jul_pval < 0.05, jul_img, np.nan)
aug_msk = np.where(aug_pval < 0.05, aug_img, np.nan)
sep_msk = np.where(sep_pval < 0.05, sep_img, np.nan)
oct_msk = np.where(oct_pval < 0.05, oct_img, np.nan)
nov_msk = np.where(nov_pval < 0.05, nov_img, np.nan)
dec_msk = np.where(dec_pval < 0.05, dec_img, np.nan)

significant_results = np.stack([jan_msk, feb_msk, mar_msk, apr_msk, may_msk, jun_msk, jul_msk, aug_msk, sep_msk, oct_msk, nov_msk, dec_msk], axis=0)

output_path = "/home/sirsimsius/Dokumente/Arbeit/KU/GEMLST/GIS/Output_Trends/lst_trend_monthly_seasonal_significant.tif"
with rio.open(output_path, 'w', driver='GTiff', height=significant_results.shape[1], width=significant_results.shape[2], count=12, dtype=significant_results.dtype, crs=src.crs, transform=src.transform) as dst:
    for i in range(12):
        dst.write(significant_results[i, :, :], i + 1)

```

```python
# Split LST quarterly and find LST trends with p<0.05.

path = "/home/sirsimsius/Dokumente/Arbeit/KU/GIS/Output_Trends/quarterly/lst_trend_4seasons_quarters.tif"

with rio.open(path) as src:
    mam_img = src.read(1)
    jja_img = src.read(2)
    son_img = src.read(3)
    djf_img = src.read(4)

    mam_pval = src.read(5)
    jja_pval = src.read(6)
    son_pval = src.read(7)
    djf_pval = src.read(8)

    crs = src.crs
    transform = src.transform

mam_msk = np.where(mam_pval < 0.05, mam_img, np.nan)
jja_msk = np.where(jja_pval < 0.05, jja_img, np.nan)
son_msk = np.where(son_pval < 0.05, son_img, np.nan)
djf_msk = np.where(djf_pval < 0.05, djf_img, np.nan)

significant_results = np.stack([mam_msk, jja_msk, son_msk, djf_msk])
band_names = ["mam", "jja", "son", "djf"]

output_path = "/home/sirsimsius/Dokumente/Arbeit/KU/GIS/Output_Trends/quarterly/lst_trend_4seasons_quarters_significant.tif"

with rio.open(
    output_path,
    "w",
    driver="GTiff",
    height=significant_results.shape[1],
    width=significant_results.shape[2],
    count=4,
    dtype=significant_results.dtype,
    crs=crs,
    transform=transform,
) as dst:
    for i, name in enumerate(band_names):
        dst.write(significant_results[i], i + 1)
        dst.set_band_description(i + 1, name)

```

```python
# Annual LST trends with p<0.05.

path = "/home/sirsimsius/Dokumente/Arbeit/KU/GIS/Output_Trends/quarterly/lst_trend_4seasons.tif"
with rio.open(path) as src:

    lin_slope_yr = src.read(1)
    lin_intercept_yer = src.read(2)
    lin_pval = src.read(3)
    tau_img = src.read(4)
    pval_img = src.read(5)
    senslope_yr = src.read(6)

    crs = src.crs
    transform = src.transform

tau_sig = np.where(pval_img < 0.05, tau_img, np.nan)
lin_slope_yr_sig = np.where(lin_pval < 0.05, lin_slope_yr, np.nan)
lin_intercept_yer_sig = np.where(lin_pval < 0.05, lin_intercept_yer, np.nan)
senslope_yr_sig = np.where(pval_img < 0.05, senslope_yr, np.nan)

significant_results = np.stack([tau_sig, lin_slope_yr_sig, lin_intercept_yer_sig, senslope_yr_sig])
band_names = ["tau", "lin_slope_yr", "lin_intercept_yer", "senslope_yr"]

output_path = "/home/sirsimsius/Dokumente/Arbeit/KU/GIS/Output_Trends/quarterly/lst_trend_4seasons_significant.tif"

with rio.open(
    output_path,
    "w",
    driver="GTiff",
    height=significant_results.shape[1],
    width=significant_results.shape[2],
    count=4,
    dtype=significant_results.dtype,
    crs=crs,
    transform=transform,
) as dst:
    for i, name in enumerate(band_names):
        dst.write(significant_results[i], i + 1)
        dst.set_band_description(i + 1, name)
```
